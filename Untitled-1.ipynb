```python
import serial
import numpy as np
import time
import matplotlib.pyplot as plt
from IPython.display import clear_output, display

plt.style.use('dark_background')
```

```python
def _readline(ser: serial.Serial, timeout: int):
    _t0 = time.time()
    s = ""
    while (time.time() - _t0 < timeout):
        if ser.in_waiting > 0:
            c = ser.read()
            if (c > bytes([0])) and (c < bytes([128])):
                if (c == b'\n') or (c == b'\r'):
                    if s == "":
                        pass
                    elif len(s) > 2:
                        return s
                else:
                    s += c.decode()
                
    return s
```

```python
with serial.Serial("COM9", 115200, rtscts=False) as ser:
    ser.write(b"hello\n")
    print(_readline(ser, .25))
    print(_readline(ser, .5))
```

```python
with serial.Serial("COM9", 115200, rtscts=False) as ser:
    ser.write(b"ww\n")
    s = _readline(ser, .5)
    print(s)
    s = _readline(ser, .5)
    print(s)

```
